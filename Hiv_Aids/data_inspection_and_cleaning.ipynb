#### convert all columns to lowercase

```python
df.columns = df.columns.str.lower().str.strip().str.replace(' ','_')
```

#### convert all non numeric rows to lower case

```python
non_num = df.select_dtypes('object').columns
for col in non_num:
    df[col] = df[col].str.lower()
```
